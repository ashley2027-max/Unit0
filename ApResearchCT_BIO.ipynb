### Graphing


Graphs and plots allow us to communicate trends in data clearly, typically catering towards wider audiences (why might this be?). In WL, there are several ways to plot data. **Using the dataset and your chosen variables, your task is to determine which of the plots below communicates trends in your data the best** (i.e. which graphs work, and which give you an error).

```r
data <- penguins

scatter_plot(data = penguins,
             x = "island",
             y = "body_mass_g")
```

```r
density_plot(data= penguins%>%filter(island!="NA"),
             x = body_mass_g,
             color_by=island)
```

```r
boxplot(body_mass_g~island, data = penguins,
       xlab = "Island", ylab = "Body mass (g)", main = "Body Mass by Island")
```

Now that we've got some visualizations for our data, it is time to use statistics to make empirical conclusions in regards to our research question.

*If you are trying to understand if one variable (like bill length; my predictor) can predict another (like bill depth; my outcome), use the code and corresponding output below.*

### Analysis of Variance (ANOVA): Is there a *difference* between these groups of things?


*If you are trying to understand if 3 or more groups (species) of associations (penguins) are statistically different by some quantity (bill length, bill depth, etc), use the code and corresponding output below.*
